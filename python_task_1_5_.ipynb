```python
# Task 1: Cinema Ticket Price
user_name = input ("Enter you name: ")
age = int (input ("Enter you age: "))

if age < 5 or age > 100:
  ticket_price = 0

elif age >= 5 and age <= 12 :
  ticket_price = 300

elif age >= 13 and age <= 59:
  ticket_price = 700

else :
  ticket_price = 400

print("Customer Name: " , user_name)
print("Ticket Price: " "Rs: " , ticket_price)
```

```python
#Task 2: Temperature Converter

Celsius_temperature = int (input ("Enter Celsius temperature: "))

Fahrenheit = (Celsius_temperature  * 9/5) + 32

if Celsius_temperature <10 :
    whether_condition = "Very Cold"

elif Celsius_temperature >= 10 and Celsius_temperature <= 19 :
  whether_condition ="Cold"

elif Celsius_temperature >= 20 and Celsius_temperature <= 29 :
  whether_condition ="Normal"
else:
  whether_condition ="Hot"

print(whether_condition)
print(Fahrenheit)
```

```python
#Task 4: Weekly Steps Tracker

total_steps = 0

for i in range(1, 8):
    user_input = int(input("Enter your steps: "))
    total_steps = total_steps + user_input

average_step = total_steps / 7

print("Total Steps:", total_steps)
print("Average Steps:", average_step)

if average_step >= 5000:

    print("Active Week")
else:
    print("Needs More Activity")
```

```python
#Task 5: Secure PIN Checker

secret_pin = 1234

for i in range(1, 4):

    user_input = int(input("Enter your secret PIN: "))

    if user_input == secret_pin:
        print("Access Granted")
        break

    else:
        remaining_attempts = 3 - i
        print("Access Denied")
        print("Remaining attempts:", remaining_attempts)

        if i == 3:
            print("Account Blocked")
```
