```python
#Advanced-Level Tasks
#Task 11: Personal Expense Tracker

expenses = []


# Add Expenses
def add_expenses():

    total_expenses = int(input("Enter total number of expenses: "))

    for i in range(total_expenses):

        print("\nEnter details for Expense", i + 1)

        expense_name = input("Enter expense name: ")

        category = input(
            "Enter category (Food/Transport/Shopping/Bills/Other): "
        ).title()

        amount = float(input("Enter amount: "))

        expense = {
            "name": expense_name,
            "category": category,
            "amount": amount
        }

        expenses.append(expense)

    print("\nAll expenses added successfully!")


# Calculate Total Spending
def calculate_total():

    total = 0

    for expense in expenses:
        total = total + expense["amount"]

    return total


# Calculate Category Spending
def category_spending():

    food = 0
    transport = 0
    shopping = 0
    bills = 0
    other = 0

    for expense in expenses:

        if expense["category"] == "Food":
            food = food + expense["amount"]

        elif expense["category"] == "Transport":
            transport = transport + expense["amount"]

        elif expense["category"] == "Shopping":
            shopping = shopping + expense["amount"]

        elif expense["category"] == "Bills":
            bills = bills + expense["amount"]

        else:
            other = other + expense["amount"]

    print("\n--- Category Spending ---")
    print("Food:", food)
    print("Transport:", transport)
    print("Shopping:", shopping)
    print("Bills:", bills)
    print("Other:", other)


# Find Highest Expense
def highest_expense():

    if len(expenses) == 0:
        return None

    highest = expenses[0]

    for expense in expenses:

        if expense["amount"] > highest["amount"]:
            highest = expense

    return highest


# Final Report
def final_report():

    total = calculate_total()

    highest = highest_expense()

    print("\n========== EXPENSE REPORT ==========")

    print("Total Spending: Rs.", total)

    category_spending()

    if highest != None:
        print("\n--- Highest Expense ---")
        print("Name:", highest["name"])
        print("Category:", highest["category"])
        print("Amount: Rs.", highest["amount"])

    budget = float(input("\nEnter your budget: "))

    if total > budget:
        print("WARNING: Your budget has been exceeded!")
        print("Exceeded Amount: Rs.", total - budget)

    else:
        print("You are within your budget.")
        print("Remaining Budget: Rs.", budget - total)

    print("====================================")


# Main Program
add_expenses()

final_report()
```

```python
#Task 12: Store Inventory Manager


products = []


# 1. Add Product
def add_product():

    product_name = input("Enter Product Name: ")

    # Check if product already exists
    for product in products:

        if product["name"].lower() == product_name.lower():
            print("Product already exists!")
            return

    price = float(input("Enter Product Price: "))
    quantity = int(input("Enter Available Quantity: "))

    product = {
        "name": product_name,
        "price": price,
        "quantity": quantity
    }

    products.append(product)

    print("Product added successfully!")


# 2. Sell Product
def sell_product():

    product_name = input("Enter Product Name to sell: ")

    for product in products:

        if product["name"].lower() == product_name.lower():

            quantity = int(input("Enter quantity to sell: "))

            if quantity <= product["quantity"]:
                product["quantity"] = product["quantity"] - quantity

                print("Product sold successfully!")
                print("Remaining Stock:", product["quantity"])

            else:
                print("Insufficient stock!")
                print("Available Quantity:", product["quantity"])

            return

    print("Product not found!")


# 3. Restock Product
def restock_product():

    product_name = input("Enter Product Name to restock: ")

    for product in products:

        if product["name"].lower() == product_name.lower():

            quantity = int(input("Enter quantity to add: "))

            product["quantity"] = product["quantity"] + quantity

            print("Product restocked successfully!")
            print("New Quantity:", product["quantity"])

            return

    print("Product not found!")


# 4. Search Product
def search_product():

    product_name = input("Enter Product Name to search: ")

    for product in products:

        if product["name"].lower() == product_name.lower():

            print("\nProduct Found!")
            print("Product Name:", product["name"])
            print("Price: Rs.", product["price"])
            print("Quantity:", product["quantity"])

            if product["quantity"] < 5:
                print("Status: Low Stock")
            else:
                print("Status: Stock Available")

            return

    print("Product not found!")


# 5. Display Inventory
def display_inventory():

    if len(products) == 0:
        print("No products available.")
        return

    print("\n========== Inventory ==========")

    for product in products:

        print("Product Name:", product["name"])
        print("Price: Rs.", product["price"])
        print("Quantity:", product["quantity"])

        if product["quantity"] < 5:
            print("Status: LOW STOCK")
        else:
            print("Status: Available")

        print("-----------------------------")


# 6. Display Low Stock Products
def low_stock():

    print("\n========== Low Stock Products ==========")

    found = False

    for product in products:

        if product["quantity"] < 5:

            print("Product Name:", product["name"])
            print("Price: Rs.", product["price"])
            print("Quantity:", product["quantity"])
            print("-----------------------------")

            found = True

    if found == False:
        print("No low-stock products.")


# 7. Calculate Total Inventory Value
def inventory_value():

    total_value = 0

    for product in products:

        product_value = product["price"] * product["quantity"]

        total_value = total_value + product_value

    print("\nTotal Inventory Value: Rs.", total_value)


# Main Menu
while True:

    print("\n========== STORE INVENTORY MANAGER ==========")

    print("1. Add Product")
    print("2. Sell Product")
    print("3. Restock Product")
    print("4. Search Product")
    print("5. Display Inventory")
    print("6. Display Low Stock Products")
    print("7. Total Inventory Value")
    print("8. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        add_product()

    elif choice == "2":
        sell_product()

    elif choice == "3":
        restock_product()

    elif choice == "4":
        search_product()

    elif choice == "5":
        display_inventory()

    elif choice == "6":
        low_stock()

    elif choice == "7":
        inventory_value()

    elif choice == "8":
        print("Thank you for using Store Inventory Manager!")
        break

    else:
        print("Invalid choice! Please try again.")
```

```python
#Task 13: Library Book Management System

books = []


# 1. Add Book
def add_book():

    book_id = int(input("Enter Book ID: "))

    # Check duplicate Book ID
    for book in books:

        if book["id"] == book_id:
            print("Book ID already exists!")
            return

    title = input("Enter Book Title: ")
    author = input("Enter Author Name: ")

    book = {
        "id": book_id,
        "title": title,
        "author": author,
        "available": True
    }

    books.append(book)

    print("Book added successfully!")


# 2. Display Books
def display_books():

    if len(books) == 0:
        print("No books available in the library.")
        return

    print("\n========== Library Books ==========")

    for book in books:

        if book["available"]:
            status = "Available"
        else:
            status = "Borrowed"

        print("Book ID:", book["id"])
        print("Title:", book["title"])
        print("Author:", book["author"])
        print("Status:", status)
        print("--------------------------------")


# 3. Search Book
def search_book():

    title = input("Enter book title to search: ")

    for book in books:

        if book["title"].lower() == title.lower():

            print("\nBook Found!")
            print("Book ID:", book["id"])
            print("Title:", book["title"])
            print("Author:", book["author"])

            if book["available"]:
                print("Status: Available")
            else:
                print("Status: Borrowed")

            return

    print("Book not found!")


# 4. Borrow Book
def borrow_book():

    book_id = int(input("Enter Book ID to borrow: "))

    for book in books:

        if book["id"] == book_id:

            if book["available"]:
                book["available"] = False
                print("Book borrowed successfully!")
            else:
                print("This book is already borrowed.")

            return

    print("Book not found!")


# 5. Return Book
def return_book():

    book_id = int(input("Enter Book ID to return: "))

    for book in books:

        if book["id"] == book_id:

            if book["available"]:
                print("This book is already available.")
            else:
                book["available"] = True
                print("Book returned successfully!")

            return

    print("Book not found!")


# Main Menu
while True:

    print("\n========== Library Management System ==========")

    print("1. Add Book")
    print("2. Display Books")
    print("3. Search Book")
    print("4. Borrow Book")
    print("5. Return Book")
    print("6. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        add_book()

    elif choice == "2":
        display_books()

    elif choice == "3":
        search_book()

    elif choice == "4":
        borrow_book()

    elif choice == "5":
        return_book()

    elif choice == "6":
        print("Thank you for using Library Management System!")
        break

    else:
        print("Invalid choice! Please try again.")
```

```python
#Task 14: Restaurant Ordering System

menu = {
    "Burger": 500,
    "Pizza": 1200,
    "Pasta": 800,
    "Drink": 200
}

orders = []


# Display Menu
print("========= Restaurant Menu ==========")

for key, value in menu.items():
    print(f"{key} : Rs. {value}")


# Take Orders
while True:

    item_name = input("Enter your item name: ").title()

    if item_name in menu:

        quantity = int(input("Enter quantity of your item: "))

        price = menu[item_name]

        sub_total = price * quantity

        orders.append({
            "item": item_name,
            "quantity": quantity,
            "subtotal": sub_total
        })

        print("Item added successfully!")

    else:
        print("This item is not available.")

    choice = input("You want to add more items? yes/no: ").lower()

    if choice == "no":
        break


# Calculate Total
total = 0

for order in orders:
    total = total + order["subtotal"]


# Calculate Tax
tax = total * 0.13


# Calculate Discount
if total >= 3000:
    discount = total * 0.10
else:
    discount = 0


# Calculate Final Bill
grand_total = total + tax - discount


# Receipt Function
def receipt():

    print("\n======== Receipt ========")

    for order in orders:
        print(
            f'{order["item"]} x {order["quantity"]} = '
            f'Rs. {order["subtotal"]}'
        )

    print("-------------------------")
    print("Total: Rs.", total)
    print("Tax 13%: Rs.", tax)
    print("Discount: Rs.", discount)
    print("-------------------------")
    print("Final Bill: Rs.", grand_total)
    print("=========================")


receipt()

```

```python
#Task 15: Clinic Appointment Manager
patients = []


# Add Appointment
def add_patient():

    patient_id = int(input("Enter Patient ID: "))

    # Duplicate ID check
    for patient in patients:
        if patient["id"] == patient_id:
            print("Patient ID already exists!")
            return

    patient_name = input("Enter Patient Name: ")
    patient_age = int(input("Enter Patient Age: "))
    problem = input("Enter Disease/Problem: ")
    appointment_time = input("Enter Appointment Time: ")
    emergency = input("Is this an emergency patient? (yes/no): ").lower()

    # Emergency validation
    while emergency != "yes" and emergency != "no":
        print("Please enter yes or no.")
        emergency = input("Is this an emergency patient? (yes/no): ").lower()

    patient = {
        "id": patient_id,
        "name": patient_name,
        "age": patient_age,
        "problem": problem,
        "time": appointment_time,
        "emergency": emergency
    }

    patients.append(patient)

    print("Appointment added successfully!")


# Search Patient
def search_patient():

    patient_id = int(input("Enter Patient ID to search: "))

    for patient in patients:

        if patient["id"] == patient_id:

            print("\nPatient Found!")
            print("Patient ID:", patient["id"])
            print("Patient Name:", patient["name"])
            print("Age:", patient["age"])
            print("Problem:", patient["problem"])
            print("Appointment Time:", patient["time"])
            print("Emergency:", patient["emergency"])

            return

    print("Patient not found!")


# Display Appointments
def display_appointments():

    if len(patients) == 0:
        print("No appointments available.")
        return

    print("\n--- Emergency Patients ---")

    for patient in patients:

        if patient["emergency"] == "yes":

            print("ID:", patient["id"])
            print("Name:", patient["name"])
            print("Age:", patient["age"])
            print("Problem:", patient["problem"])
            print("Time:", patient["time"])
            print("Emergency:", patient["emergency"])
            print("------------------------")


    print("\n--- Normal Patients ---")

    for patient in patients:

        if patient["emergency"] == "no":

            print("ID:", patient["id"])
            print("Name:", patient["name"])
            print("Age:", patient["age"])
            print("Problem:", patient["problem"])
            print("Time:", patient["time"])
            print("Emergency:", patient["emergency"])
            print("------------------------")


# Cancel Appointment
def cancel_appointment():

    patient_id = int(input("Enter Patient ID to cancel: "))

    for patient in patients:

        if patient["id"] == patient_id:

            patients.remove(patient)

            print("Appointment cancelled successfully!")
            return

    print("Patient not found!")


# Clinic Summary
def clinic_summary():

    emergency_count = 0
    normal_count = 0

    for patient in patients:

        if patient["emergency"] == "yes":
            emergency_count = emergency_count + 1

        else:
            normal_count = normal_count + 1

    print("\n--- Clinic Summary ---")
    print("Total Patients:", len(patients))
    print("Emergency Patients:", emergency_count)
    print("Normal Patients:", normal_count)


# Main Menu
while True:

    print("\n==============================")
    print("   CLINIC APPOINTMENT MANAGER")
    print("==============================")

    print("1. Add Appointment")
    print("2. Search Patient")
    print("3. Display Appointments")
    print("4. Cancel Appointment")
    print("5. Display Clinic Summary")
    print("6. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        add_patient()

    elif choice == "2":
        search_patient()

    elif choice == "3":
        display_appointments()

    elif choice == "4":
        cancel_appointment()

    elif choice == "5":
        clinic_summary()

    elif choice == "6":
        print("Thank you for using Clinic Appointment Manager!")
        break

    else:
        print("Invalid choice! Please try again.")
```
