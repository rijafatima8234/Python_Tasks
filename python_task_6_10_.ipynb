```python
# Task 6: Grocery Shopping Bill

total_products = int(input("Enter total products: "))

total_bill = 0

for i in range(total_products):
    product_name = input("Enter product name: ")
    price = int(input("Enter your Price: "))
    quantity = int(input("Enter your Quantity: "))

    Subtotal = price * quantity

    total_bill = total_bill + Subtotal

if total_bill >= 5000:
    Discount = total_bill * 10 / 100
    Final_Bill = total_bill - Discount
else:
    Discount = 0
    Final_Bill = total_bill

print("Total Bill:", total_bill)
print("Discount:", Discount)
print("Final Bill:", Final_Bill)
```

```python
#Task 7: Employee Weekly Salary

employee_name = input("Enter employee name: ")
working_Hours = int(input("Enter total Working Hour: "))
hourly_rate = int(input("Enter hourly rate: "))

if working_Hours <= 40:
    Normal_salary = working_Hours * hourly_rate
    total_salary = Normal_salary

else:
    Normal_salary = 40 * hourly_rate

    overtime_hours = working_Hours - 40
    Overtime_rate = hourly_rate * 1.5
    Overtime_Salary = overtime_hours * Overtime_rate

    total_salary = Normal_salary + Overtime_Salary

print("Employee Name:", employee_name)
print("Total Weekly Salary:", total_salary)
```

```python
#Task 10: Contact Book

contacts = {}

while True:
    print("\n--- Contact Book ---")
    print("1. Add Contact")
    print("2. Search Contact")
    print("3. Display All Contacts")
    print("4. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        name = input("Enter contact name: ").title()

        if name in contacts:
            print("Contact already exists!")
        else:
            phone = input("Enter phone number: ")
            contacts[name] = phone
            print("Contact added successfully!")

    elif choice == "2":
        name = input("Enter contact name to search: ").title()

        if name in contacts:
            print("Name:", name)
            print("Phone:", contacts[name])
        else:
            print("Contact not found!")

    elif choice == "3":
        if len(contacts) == 0:
            print("No contacts available.")
        else:
            print("\n--- All Contacts ---")
            for name, phone in contacts.items():
                print("Name:", name, "| Phone:", phone)

    elif choice == "4":
        print("Exiting Contact Book. Goodbye!")
        break
    else:
        print("Invalid choice! Please select 1-4.")

```
